```python
# Importing Liabs

import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('image.jpg')

gray = cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)

plt.imshow(gray,cmap= 'gray')
plt.title("Aquired Image to Grayscale")
plt.axis('off')
```

```python
sampled = cv2.resize(gray,(120,80))

plt.subplot(1,2,1)
plt.imshow(gray,cmap= 'gray')
plt.title("Grayscale")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(sampled,cmap= 'gray')
plt.title("Sampled")
plt.axis('off')
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.hist(sampled.ravel(),bins = 256)
plt.title("Sampled Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(1,2,2)
plt.hist(gray.ravel(),bins = 256)
plt.title("Grayscale Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
```

```python
equalized = cv2.equalizeHist(gray)

plt.figure(figsize=(12,4))
plt.hist(equalized.ravel(),bins = 256)
plt.title("equalized Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
```

```python
plt.figure(figsize = (16,6))

plt.subplot(1,3,1)
plt.imshow(gray,cmap = 'gray')
plt.title("Grayscale")

plt.subplot(1,3,2)
plt.imshow(sampled,cmap = 'gray')
plt.title("Sampled")

plt.subplot(1,3,3)
plt.imshow(equalized,cmap= 'gray')
plt.title("Equalized")
```
